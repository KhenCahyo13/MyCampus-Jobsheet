## **PRACTICUM - Importing Library in Python**

```python
import numpy as np
import pandas as pd
import cv2 as cv
import matplotlib.pylab as plt
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image

print('Successfully imported libraries');
```

## **PRACTICUM - Reading & Showing Image in Python**

```python
image_urls = [
    "https://images.squarespace-cdn.com/content/v1/607f89e638219e13eee71b1e/1684821560422-SD5V37BAG28BURTLIXUQ/michael-sum-LEpfefQf4rU-unsplash.jpg",
    "https://dogagingproject.org/_next/image?url=https%3A%2F%2Fcontent.dogagingproject.org%2Fwp-content%2Fuploads%2F2020%2F11%2Fhelena-lopes-S3TPJCOIRoo-unsplash-scaled.jpg&w=3840&q=75",
    "https://blog.nature.org/wp-content/uploads/2020/02/29555193323_15d785590f_k.jpg?w=1024"
];

# Loop each urls and do some actions
for url in image_urls:
  # Showing Image
  image = io.imread(url)
  image = cv.resize(image, (0, 0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = cv.hconcat((image, image_2))
  cv2_imshow(final_frame)
  print('\n');

  # Print Image Size
  height = image_2.shape[0]
  width = image_2.shape[1]
  print('Image resolution: height * width = ', height, '*', width);
  print('\n')

  # Drawing White Horizontal Line
  image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  for y in range (width):
    image_3[int((height) / 2), y] = [255, 255, 255]

  final_frame2 = cv.hconcat((image_2, image_3))
  cv2_imshow(final_frame2)
  print('\n');
```
