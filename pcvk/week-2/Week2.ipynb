# Practicum

## Step 1 - Import Libraries & Define the Data

```python
import cv2 as cv
import matplotlib.pyplot as plt
import numpy as np

from skimage import io

# Image Address
image_url = 'https://asset.kompas.com/crops/__wbPDdjsiDhfvS0lHSSp9Hpyig=/0x0:800x533/1200x800/data/photo/2021/09/19/614708b006de2.jpg';
```

## Step 2 - Image Processing

```python
# Default Image
default_image = io.imread(image_url)
plt.imshow(default_image);
```

```python
# Convert to BGR
bgr_image = cv.cvtColor(default_image, cv.COLOR_RGB2BGR)
plt.imshow(bgr_image);
```

```python
# Grayscale & Magma Image
gray_image = cv.cvtColor(bgr_image, cv.COLOR_BGR2GRAY)
# plt.imshow(gray_image, cmap='gray');
plt.imshow(gray_image, cmap='magma');
```

```python
# RGB Resized to 512 x 1024
resized_image = cv.resize(default_image, (512, 1024), interpolation=cv.INTER_CUBIC)
plt.imshow(resized_image);
```

```python
# Flip Image
flipped_image = cv.flip(default_image, -1)
figure = plt.figure(figsize=(10, 10))
ax = figure.add_subplot(111)
ax.imshow(flipped_image)
```

## Step 3 - Black Image (INT16)

```python
black_img16 = np.zeros((512, 512, 3), dtype=np.int16)
plt.imshow(black_img16)
```

```python
# Drawing Rectangle
cv.rectangle(black_img16, (384, 0), (510, 150), (0, 255, 0), thickness=10)
plt.imshow(black_img16)
```

```python
# Drawing Square Shape
cv.rectangle(black_img16, (200, 200), (300, 300), (0, 0, 255), thickness=15)
plt.imshow(black_img16)
```

```python
# Drawing Circle Shape
cv.circle(black_img16, (100, 100), 50, (255, 0, 0), thickness=8)
plt.imshow(black_img16)
```

```python
# Drawing Line
cv.line(black_img16, (0, 0), (512, 512), (255, 255, 255), thickness=5)
plt.imshow(black_img16)
```

```python
# Adding Text
font = cv.FONT_HERSHEY_COMPLEX
cv.putText(black_img16, 'World', (10, 500), font, 4, (255, 255, 0), thickness=2)
plt.imshow(black_img16)
```

## Step 3 - Black Image (INT32)

```python
vertices = np.array([[100, 300], [200, 200], [400, 300], [200, 400]], dtype=np.int32)
```

```python
# Draw the Shape
black_img32 = np.zeros((512, 512, 3), dtype=np.int16)
cv.polylines(black_img32, [vertices], isClosed=True, color=(255, 0, 0), thickness=5)
plt.imshow(black_img32)
```
